# Capítulo 11: El Espectro de los Algoritmos

Comparación de 4 algoritmos de optimización para TSP:
1. Fuerza Bruta (Exacto)
2. Vecino Más Cercano (Greedy)
3. 2-opt (Búsqueda Local)
4. Recocido Simulado (Búsqueda Estocástica Global)

```python
import sys
sys.path.append('..')
from src.experiments.algorithm_comparison import (
    generar_instancia, matriz_distancias,
    fuerza_bruta, vecino_mas_cercano,
    dos_opt, recocido_simulado
)
import time
```

```python
# Instancia aleatoria
ciudades = generar_instancia(8)
matriz = matriz_distancias(ciudades)

print("Coordenadas:")
for i, (x, y) in enumerate(ciudades):
    print(f"  {i}: ({x:.2f}, {y:.2f})")
```

```python
t0 = time.time()
ruta_fb, costo_fb = fuerza_bruta(matriz)
t_fb = time.time() - t0
print(f"Fuerza Bruta: costo={costo_fb:.4f}, tiempo={t_fb:.6f}s")

t0 = time.time()
ruta_g, costo_g = vecino_mas_cercano(matriz)
t_g = time.time() - t0
print(f"Greedy: costo={costo_g:.4f}, diff={costo_g-costo_fb:.4f}, tiempo={t_g:.6f}s")

t0 = time.time()
ruta_2opt, costo_2opt = dos_opt(matriz, ruta_g[:-1])
t_2opt = time.time() - t0
print(f"2-opt: costo={costo_2opt:.4f}, diff={costo_2opt-costo_fb:.4f}, tiempo={t_2opt:.6f}s")

t0 = time.time()
ruta_sa, costo_sa = recocido_simulado(matriz, iteraciones=20000)
t_sa = time.time() - t0
print(f"Recocido: costo={costo_sa:.4f}, diff={costo_sa-costo_fb:.4f}, tiempo={t_sa:.6f}s")
```
