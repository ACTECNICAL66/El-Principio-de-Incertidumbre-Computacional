# Capítulo 10: El Laboratorio de la "Trampa"

Experimento controlado que demuestra el Principio de Incertidumbre Computacional mediante la instancia adversarial "Trampa V2.0" del Problema del Viajante (TSP).

**Objetivo**: Mostrar cómo un algoritmo greedy (Vecino Más Cercano), limitado a información local, es engañado por una instancia que maximiza la entropía residual en cada punto de decisión.

```python
import sys
sys.path.append('..')
from src.experiments.trap_v2 import (
    construir_matriz_trampa_v2,
    algoritmo_vecino_mas_cercano,
    calcular_costo,
    fuerza_bruta_tsp
)
```

```python
ciudades, matriz = construir_matriz_trampa_v2()
print("Matriz de distancias:")
for c in ciudades:
    print(f"  {c}: {[matriz[c][d] for d in ciudades]}")
```

```python
ruta_greedy = algoritmo_vecino_mas_cercano(matriz, 'A')
costo_greedy = calcular_costo(matriz, ruta_greedy)
print(f"Ruta greedy: {' -> '.join(ruta_greedy)}")
print(f"Costo greedy: {costo_greedy}")
```

```python
ruta_opt, costo_opt = fuerza_bruta_tsp(matriz)
print(f"Ruta óptima: {' -> '.join(ruta_opt)}")
print(f"Costo óptimo: {costo_opt}")
print(f"Ratio greedy/óptimo: {costo_greedy / costo_opt:.1f}x")
```

## Conclusión

El algoritmo greedy, tomando decisiones localmente óptimas, produce un tour 7.9 veces peor que el óptimo global. Esto valida el Principio de Incertidumbre Computacional: la información local es insuficiente para garantizar optimalidad global.
