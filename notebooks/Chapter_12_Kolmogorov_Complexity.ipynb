# Capítulo 12: Complejidad Descriptiva y Kolmogórov

Estimación de la complejidad de Kolmogórov de rutas del TSP mediante compresión con zlib.

**Pregunta**: ¿Tienen las soluciones óptimas una estructura simple que permita atajos de búsqueda?

```python
import sys
sys.path.append('..')
from src.experiments.kolmogorov_complexity import (
    complejidad_aproximada,
    generar_ruta_aleatoria,
    generar_instancia_compleja
)
import json
```

```python
# Comparar ruta óptima vs aleatoria
ruta_opt = [0, 3, 2, 7, 5, 6, 1, 4, 0]
ruta_rand = generar_ruta_aleatoria(8)

print(f"Ruta óptima: {ruta_opt}")
print(f"  Tamaño original: {len(json.dumps(ruta_opt).encode('utf-8'))} bytes")
print(f"  Comprimido (K aprox): {complejidad_aproximada(ruta_opt)} bytes")

print(f"\nRuta aleatoria: {ruta_rand}")
print(f"  Tamaño original: {len(json.dumps(ruta_rand).encode('utf-8'))} bytes")
print(f"  Comprimido (K aprox): {complejidad_aproximada(ruta_rand)} bytes")
```

```python
# Escalar a instancias más grandes
print("Escalando a n más grandes:")
for n in [10, 20, 50, 100, 200]:
    rc = generar_instancia_compleja(n)
    comp = complejidad_aproximada(rc)
    raw = len(json.dumps(rc).encode('utf-8'))
    print(f"  n={n:3d} | original: {raw:4d} bytes | comprimido: {comp:4d} bytes")
```

## Conclusión

Las rutas óptimas pueden ser arbitrariamente complejas (Teorema 19.1). No hay un patrón simple universal que delate a la solución óptima, reforzando el PIC: la información sobre la optimalidad está codificada globalmente.
